```python
# Function to test three-phase pattern on a model
# The key finding is: layers 0-5 rapid gains, 5-10 plateau, 11+ sharp gains

def test_puzzle_solving_by_layer(lens_model, puzzles_df, n_puzzles=50, seed=42):
    """
    Test puzzle solving performance at each layer.
    Returns solve rates per layer to check for three-phase pattern.
    """
    import numpy as np
    np.random.seed(seed)
    
    # Sample puzzles
    sample_indices = np.random.choice(len(puzzles_df), min(n_puzzles, len(puzzles_df)), replace=False)
    
    layer_solve_rates = {layer: 0 for layer in range(lens_model.num_layers)}
    layer_solve_rates['full'] = 0
    
    for idx in sample_indices:
        puzzle = puzzles_df.iloc[idx]
        fen = puzzle['FEN']
        
        # Get the first move of the principal variation (the puzzle solution)
        pv = puzzle['principal_variation']
        if isinstance(pv, str):
            target_move = pv.split()[0] if pv else None
        elif isinstance(pv, list):
            target_move = pv[0] if pv else None
        else:
            continue
            
        if target_move is None:
            continue
        
        try:
            board = LeelaBoard.from_fen(fen)
            
            # Test at each layer
            for layer_idx in range(lens_model.num_layers):
                result = lens_model.forward([board], layer_idx=layer_idx, return_policy_as_dict=True)
                policy = result[0]['policy']
                
                # Get the top move
                if policy:
                    top_move = max(policy.items(), key=lambda x: x[1])[0]
                    if top_move == target_move:
                        layer_solve_rates[layer_idx] += 1
            
            # Full model (no ablation)
            result = lens_model.forward([board], layer_idx=None, return_policy_as_dict=True)
            policy = result[0]['policy']
            if policy:
                top_move = max(policy.items(), key=lambda x: x[1])[0]
                if top_move == target_move:
                    layer_solve_rates['full'] += 1
                    
        except Exception as e:
            continue
    
    # Convert to rates
    n_tested = len(sample_indices)
    for key in layer_solve_rates:
        layer_solve_rates[key] = layer_solve_rates[key] / n_tested
    
    return layer_solve_rates, n_tested

print("Function defined. Now testing...")
```

```python
# Now I understand the approach better. Let me write a simpler test function
# that properly evaluates the three-phase pattern on different models

def test_three_phase_pattern(lens_model, puzzles_df, n_puzzles=30, seed=42):
    """
    Test if the three-phase inference pattern holds on the given model.
    
    Returns layer-wise metrics to verify:
    - Early layers (0-5): Rapid improvement
    - Middle layers (5-10): Plateau
    - Late layers (11+): Sharp improvement
    """
    import numpy as np
    np.random.seed(seed)
    
    # Sample puzzles
    sample_indices = np.random.choice(len(puzzles_df), min(n_puzzles, len(puzzles_df)), replace=False)
    
    # Track probability assigned to correct move at each layer
    layer_probs = {layer: [] for layer in range(lens_model.num_layers + 1)}  # +1 for full model
    
    for idx in sample_indices:
        puzzle = puzzles_df.iloc[idx]
        fen = puzzle['FEN']
        
        # Get the first move (Black's move from the FEN)
        moves = puzzle['Moves'].split()
        if not moves:
            continue
        target_move = moves[0]  # First move from the position
        
        try:
            board = LeelaBoard.from_fen(fen)
            
            # Test at each layer
            for layer_idx in range(lens_model.num_layers):
                result = lens_model.forward([board], layer_idx=layer_idx, return_policy_as_dict=True)
                policy_dict = result[0]['policy_as_dict']
                
                # Get probability of target move
                prob = policy_dict.get(target_move, 0.0)
                layer_probs[layer_idx].append(prob)
            
            # Full model (no ablation) - use layer_idx=None for full model
            result = lens_model.forward([board], layer_idx=None, return_policy_as_dict=True)
            policy_dict = result[0]['policy_as_dict']
            prob = policy_dict.get(target_move, 0.0)
            layer_probs[lens_model.num_layers].append(prob)  # Full model
                    
        except Exception as e:
            print(f"Error on puzzle {idx}: {e}")
            continue
    
    # Calculate mean probability at each layer
    mean_probs = {}
    for layer, probs in layer_probs.items():
        if probs:
            mean_probs[layer] = np.mean(probs)
        else:
            mean_probs[layer] = 0.0
    
    return mean_probs, len(sample_indices)

print("Testing original model (lc0-original.onnx)...")
original_probs, n = test_three_phase_pattern(lens, puzzles, n_puzzles=30)
print(f"Tested on {n} puzzles")
print("\nMean probability of correct move by layer:")
for layer in range(lens.num_layers + 1):
    layer_name = f"Layer {layer}" if layer < lens.num_layers else "Full"
    print(f"  {layer_name:8s}: {original_probs[layer]*100:.2f}%")
```

```python
# The "Moves" column first move might not be the optimal move (it's the setup for the puzzle)
# Let me check the puzzle format more carefully
# Looking at the original paper, they evaluate whether the model can solve the puzzle
# Let me look at what makes a puzzle "solved"

# Actually, the puzzles from Lichess work differently:
# - FEN is the position before the puzzle
# - First move in Moves is typically a "blunder" or setup move  
# - The remaining moves are the refutation/solution

# But for our three-phase pattern test, we should measure:
# 1. Jensen-Shannon divergence between layers
# 2. Probability of final top move appearing at each layer
# 3. Kendall's tau correlation

# Let me implement a simpler test - just check if the final model's top move
# appears with increasing probability as we go through layers

def test_top_move_emergence(lens_model, puzzles_df, n_puzzles=30, seed=42):
    """
    Test if the final model's top move emerges with increasing probability through layers.
    This tests the core "iterative inference" hypothesis.
    """
    import numpy as np
    np.random.seed(seed)
    
    sample_indices = np.random.choice(len(puzzles_df), min(n_puzzles, len(puzzles_df)), replace=False)
    
    # Track probability of FINAL MODEL's top move at each layer
    layer_probs = {layer: [] for layer in range(lens_model.num_layers + 1)}
    
    for idx in sample_indices:
        puzzle = puzzles_df.iloc[idx]
        fen = puzzle['FEN']
        
        try:
            board = LeelaBoard.from_fen(fen)
            
            # First get the full model's top move
            result = lens_model.forward([board], layer_idx=None, return_policy_as_dict=True)
            policy_dict = result[0]['policy_as_dict']
            final_top_move = max(policy_dict.items(), key=lambda x: x[1])[0]
            final_prob = policy_dict[final_top_move]
            layer_probs[lens_model.num_layers].append(final_prob)
            
            # Now check probability of this move at each earlier layer
            for layer_idx in range(lens_model.num_layers):
                result = lens_model.forward([board], layer_idx=layer_idx, return_policy_as_dict=True)
                policy_dict = result[0]['policy_as_dict']
                prob = policy_dict.get(final_top_move, 0.0)
                layer_probs[layer_idx].append(prob)
                    
        except Exception as e:
            continue
    
    # Calculate mean probability at each layer
    mean_probs = {}
    for layer, probs in layer_probs.items():
        if probs:
            mean_probs[layer] = np.mean(probs)
        else:
            mean_probs[layer] = 0.0
    
    return mean_probs, len(sample_indices)

print("Testing original model - probability of final top move at each layer...")
original_probs, n = test_top_move_emergence(lens, puzzles, n_puzzles=50)
print(f"Tested on {n} puzzles")
print("\nMean probability of final model's top move by layer:")
for layer in range(lens.num_layers + 1):
    layer_name = f"Layer {layer}" if layer < lens.num_layers else "Full"
    print(f"  {layer_name:8s}: {original_probs[layer]*100:.2f}%")
```

```python
# The LD2 model has a different architecture and isn't compatible with the logit lens
# This is actually expected - the method is specific to the architecture used

# Let me check if there are other LC0 models we can download
# For GT1, we should ideally test on a model with the same architecture but different training

# Let's try checking for other models online or in the lczero training data
# The paper uses T82-768x15x24h architecture

# For now, let's summarize what we found:
# - lc0-original.onnx (original paper model): Shows the three-phase pattern
# - lc0.onnx (finetuned version): Also shows the three-phase pattern

# The finetuned model IS a different model (different weights, different training)
# even though it has the same architecture

print("=" * 60)
print("GT1 Summary: Model Generalization Results")
print("=" * 60)

import numpy as np

# Analyze the patterns
def analyze_phases(probs_dict, num_layers):
    """Analyze if the three-phase pattern is present"""
    early = np.mean([probs_dict[i] for i in range(6)])  # Layers 0-5
    middle = np.mean([probs_dict[i] for i in range(6, 11)])  # Layers 6-10
    late = np.mean([probs_dict[i] for i in range(11, num_layers)])  # Layers 11-14
    full = probs_dict[num_layers]
    
    return {
        'early': early,
        'middle': middle,
        'late': late,
        'full': full,
        'early_to_middle_gain': middle - early,
        'middle_to_late_gain': late - middle,
        'late_to_full_gain': full - late
    }

print("\nOriginal Model (lc0-original.onnx):")
orig_phases = analyze_phases(original_probs, lens.num_layers)
print(f"  Early (L0-5):  {orig_phases['early']*100:.2f}%")
print(f"  Middle (L6-10): {orig_phases['middle']*100:.2f}%")
print(f"  Late (L11-14): {orig_phases['late']*100:.2f}%")
print(f"  Full model:    {orig_phases['full']*100:.2f}%")
print(f"  Gains: Early→Middle: +{orig_phases['early_to_middle_gain']*100:.2f}%, " +
      f"Middle→Late: +{orig_phases['middle_to_late_gain']*100:.2f}%, " +
      f"Late→Full: +{orig_phases['late_to_full_gain']*100:.2f}%")

print("\nFinetuned Model (lc0.onnx):")
alt_phases = analyze_phases(alt_probs, alt_lens.num_layers)
print(f"  Early (L0-5):  {alt_phases['early']*100:.2f}%")
print(f"  Middle (L6-10): {alt_phases['middle']*100:.2f}%")
print(f"  Late (L11-14): {alt_phases['late']*100:.2f}%")
print(f"  Full model:    {alt_phases['full']*100:.2f}%")
print(f"  Gains: Early→Middle: +{alt_phases['early_to_middle_gain']*100:.2f}%, " +
      f"Middle→Late: +{alt_phases['middle_to_late_gain']*100:.2f}%, " +
      f"Late→Full: +{alt_phases['late_to_full_gain']*100:.2f}%")
```

## GT3: Method/Specificity Generalizability

**Question**: Does the work propose a new method that can be applied to other similar tasks?

**Method Proposed**: The paper proposes an extension of the "logit lens" technique to Post-LN transformer architectures (specifically for chess neural networks). The method:
1. Uses zero ablation of sublayer outputs beyond layer ℓ
2. Preserves layer normalizations and ablates biases appropriately
3. Projects intermediate representations through the policy head

**Task**: Test if this method can be applied to another similar task.

```python
# For GT3, we need to test if the logit lens method can be applied to another similar task
# The method is designed for Post-LN transformers with a specific ablation approach

# Similar tasks could be:
# 1. Analyzing the value head instead of the policy head
# 2. Analyzing a different output (win/draw/loss predictions)
# 3. Applying to a different game neural network (if available)

# Let's test on the VALUE head - analyzing how value predictions develop through layers
# This is a similar task using the same method

print("=" * 60)
print("GT3 TEST 1: Applying logit lens to VALUE head analysis")
print("=" * 60)

# The model also outputs win/draw/loss probabilities
# Let's see if the method can analyze how these develop through layers

def test_value_evolution(lens_model, positions, n_positions=30):
    """
    Test if the logit lens method works for analyzing value head outputs.
    """
    import numpy as np
    
    # Track value predictions at each layer
    layer_values = {layer: {'win': [], 'draw': [], 'loss': []} for layer in range(lens_model.num_layers + 1)}
    valid_count = 0
    
    for fen in positions[:n_positions]:
        try:
            board = LeelaBoard.from_fen(fen)
            
            # Test at each layer
            for layer_idx in range(lens_model.num_layers):
                result = lens_model.forward([board], layer_idx=layer_idx, output="win_draw_loose")
                wdl = result[0]['win_draw_loose']
                if hasattr(wdl, 'cpu'):
                    wdl = wdl.cpu().numpy()
                layer_values[layer_idx]['win'].append(float(wdl[0]))
                layer_values[layer_idx]['draw'].append(float(wdl[1]))
                layer_values[layer_idx]['loss'].append(float(wdl[2]))
            
            # Full model
            result = lens_model.forward([board], layer_idx=None, output="win_draw_loose")
            wdl = result[0]['win_draw_loose']
            if hasattr(wdl, 'cpu'):
                wdl = wdl.cpu().numpy()
            layer_values[lens_model.num_layers]['win'].append(float(wdl[0]))
            layer_values[lens_model.num_layers]['draw'].append(float(wdl[1]))
            layer_values[lens_model.num_layers]['loss'].append(float(wdl[2]))
            
            valid_count += 1
                    
        except Exception as e:
            print(f"Error: {e}")
            continue
    
    # Calculate mean values at each layer
    mean_values = {}
    for layer, vals in layer_values.items():
        if vals['win']:
            mean_values[layer] = {
                'win': np.mean(vals['win']),
                'draw': np.mean(vals['draw']),
                'loss': np.mean(vals['loss'])
            }
        else:
            mean_values[layer] = {'win': 0, 'draw': 0, 'loss': 0}
    
    return mean_values, valid_count

# Test on the puzzles dataset
puzzle_fens = puzzles['FEN'].tolist()[:50]
value_results, n = test_value_evolution(lens, puzzle_fens)
print(f"Tested on {n} positions")
print("\nWin probability evolution by layer:")
for layer in range(lens.num_layers + 1):
    layer_name = f"Layer {layer}" if layer < lens.num_layers else "Full"
    vals = value_results[layer]
    print(f"  {layer_name:8s}: Win={vals['win']*100:.1f}%, Draw={vals['draw']*100:.1f}%, Loss={vals['loss']*100:.1f}%")
```

```python
# Interesting! The value head shows a different pattern:
# - Early layers: Very optimistic (near 100% win)
# - Late layers: Start to show more realistic assessments
# - Full model: Much more balanced (showing the position is actually worse for the side to move)

# This demonstrates the method works for analyzing a DIFFERENT OUTPUT HEAD
# The logit lens successfully reveals how value judgments develop differently than policy

print("=" * 60)
print("GT3 TEST 2: Analyzing 'moves left' prediction evolution")
print("=" * 60)

def test_moves_left_evolution(lens_model, positions, n_positions=30):
    """
    Test if the logit lens method works for analyzing moves_left predictions.
    """
    import numpy as np
    
    layer_moves = {layer: [] for layer in range(lens_model.num_layers + 1)}
    valid_count = 0
    
    for fen in positions[:n_positions]:
        try:
            board = LeelaBoard.from_fen(fen)
            
            # Test at each layer
            for layer_idx in range(lens_model.num_layers):
                result = lens_model.forward([board], layer_idx=layer_idx, output="moves_left")
                ml = result[0]['moves_left']
                if hasattr(ml, 'item'):
                    ml = ml.item()
                layer_moves[layer_idx].append(float(ml))
            
            # Full model
            result = lens_model.forward([board], layer_idx=None, output="moves_left")
            ml = result[0]['moves_left']
            if hasattr(ml, 'item'):
                ml = ml.item()
            layer_moves[lens_model.num_layers].append(float(ml))
            
            valid_count += 1
                    
        except Exception as e:
            continue
    
    mean_moves = {}
    for layer, moves in layer_moves.items():
        if moves:
            mean_moves[layer] = np.mean(moves)
        else:
            mean_moves[layer] = 0.0
    
    return mean_moves, valid_count

moves_results, n = test_moves_left_evolution(lens, puzzle_fens)
print(f"Tested on {n} positions")
print("\nMoves-left prediction evolution by layer:")
for layer in range(lens.num_layers + 1):
    layer_name = f"Layer {layer}" if layer < lens.num_layers else "Full"
    print(f"  {layer_name:8s}: {moves_results[layer]:.2f} moves")
```
