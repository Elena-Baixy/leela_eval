```python
# Setup the environment properly
import os
import subprocess
import sys

# Change to the repo directory
os.chdir('/net/scratch2/smallyan/leela_eval')

# Load environment variables
result = subprocess.run(['bash', '-c', 'source ~/.bashrc && env'], capture_output=True, text=True)
for line in result.stdout.split('\n'):
    if '=' in line and not line.startswith('_'):
        key, _, value = line.partition('=')
        os.environ[key] = value

print(f"Working directory: {os.getcwd()}")
print(f"HF_HOME: {os.environ.get('HF_HOME', 'Not set')}")
print(f"Python version: {sys.version}")
```

```python
# Need to restart to pick up the new version - but since we're in notebook let's reimport
import importlib
import sys

# Remove cached nnsight modules
mods_to_remove = [k for k in sys.modules.keys() if 'nnsight' in k]
for mod in mods_to_remove:
    del sys.modules[mod]

import nnsight
print(f"nnsight version after reinstall: {nnsight.__version__}")
```
